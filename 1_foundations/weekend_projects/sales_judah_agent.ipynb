# Project: Dating App Social Media Marketing Agents

## Business Requirements & Goals
- Build multiple AI agents that generate engaging social media content for a serious dating app
- Create specialized agents for different platforms:
  1. X Agent: Creates viral threads and engagement posts
  2. Instagram Agent: Designs post captions and story concepts and also shared pictures for posts
  3. Blog Agent: Writes SEO-optimized relationship advice articles.
- Each agent should:
  - Follow brand voice and content guidelines
  - Generate platform-specific content based on provided documents/briefs
  - Create content calendars and posting schedules
  - Track engagement metrics and adapt strategies
  - Maintain consistent messaging across platforms
## Example Use Cases
- "Generate a Twitter thread about red flags in dating"
- "Create an Instagram carousel about successful first date tips"
- "Write a blog post about 'Why traditional dating is making a comeback'"
- "Draft a week's worth of social media content around 'finding love in your 30s'"
- "Analyze which content themes perform best on each platform"

```python
from dotenv import load_dotenv
from agents import Agent, Runner, trace, function_tool
from openai.types.responses import ResponseTextDeltaEvent
from typing import Dict
import sendgrid
import os
from sendgrid.helpers.mail import Mail, Email, To, Content
import asyncio
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
instructions1 = "You are a professional sales agent working for Amoriva, \
a dating app that helps serious people find meaningful relationships. \
You write engaging content specifically for Twitter."

instructions2 = "You are a professional sales agent working for Amoriva, \
a dating app that helps serious people find meaningful and serious relationships. \
You write engaging content specifically for Instagram. You focus on engaging visuals and concise, impactful captions"

instructions3 = "You are a professional sales agent working for Amoriva, \
a dating app that helps serious people find meaningful relationships. \
You write engaging content specifically for Blog posts. You provide in-depth insights and valuable information."
```

```python
sales_agent1 = Agent(
        name="Engaging X Sales Agent",
        instructions=instructions1,
        model="gpt-4o-mini"
)

sales_agent2 = Agent(
        name="Engaging Instagram Sales Agent",
        instructions=instructions2,
        model="gpt-4o-mini"
)

sales_agent3 = Agent(
        name="Engaging Blog Sales Agent",
        instructions=instructions3,
        model="gpt-4o-mini"
)
```

```python

result = Runner.run_streamed(sales_agent1, input="Write a cold sales email")
async for event in result.stream_events():
    if event.type == "raw_response_event" and isinstance(event.data, ResponseTextDeltaEvent):
        print(event.data.delta, end="", flush=True)
```

```python
message = "Write a cold sales email"

with trace("Parallel cold emails"):
    results = await asyncio.gather(
        Runner.run(sales_agent1, message),
        Runner.run(sales_agent2, message),
        Runner.run(sales_agent3, message),
    )

outputs = [result.final_output for result in results]

for output in outputs:
    print(output + "\n\n")
```

```python
sales_picker = Agent(
    name="sales_picker",
    instructions="You pick the best and engaging sales content from the given options for social media marketing. \
Imagine you are a customer and pick the one you are most likely to respond to. \
Do not give an explanation; reply with the selected email only.",
    model="gpt-4o-mini"
)
```

```python
message = "Write the content for social media marketing for a dating app called Amoriva that helps serious people find serious and meaningful relationships."

with trace("Selection from sales people"):
    results = await asyncio.gather(
        Runner.run(sales_agent1, message),
        Runner.run(sales_agent2, message),
        Runner.run(sales_agent3, message),
    )
    outputs = [result.final_output for result in results]

    emails = "Cold sales emails:\n\n" + "\n\nEmail:\n\n".join(outputs)

    best = await Runner.run(sales_picker, emails)

    print(f"Best sales email:\n{best.final_output}")
```
