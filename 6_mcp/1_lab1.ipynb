### Let's use MCP in OpenAI Agents SDK

1. Create a Client

2. Have it spawn a server

3. Collect the tools that the server can use

Let's try the Fetch mcp-server that we looked at last week

### Special note - in case you updated the OpenAI Agents framework:

OpenAI Agents SDK recently upgraded their SDK and made a breaking change (thank you, OpenAI 😂). I haven't yet updated the version in this repo to the latest version. If you've upgraded the package yourself, then you might get an error about missing positional arguments. If you get that error, you'll need to replace `await server.list_tools()` with `await server.session.list_tools()` and change the last line from `fetch_tools` to `fetch_tools.tools`. And you'll need to make this change every time you get this error! The joys of working at the bleeding edge..

```python
# MCP CLIENT SETUP: Connect to Fetch MCP Server
# This demonstrates the core MCP architecture pattern:
# 1. MCP Client (this code) connects to MCP Server (mcp-server-fetch)
# 2. Server provides tools that our AI agent can use
# 3. Communication happens via JSON-RPC over STDIO

# Configure MCP Server parameters for the Fetch server
# "uvx" is a Python package runner that executes mcp-server-fetch
# This server provides web browsing and content fetching capabilities
fetch_params = {"command": "uvx", "args": ["mcp-server-fetch"]}

# Create MCP Client connection using STDIO transport
# MCPServerStdio manages the lifecycle of the MCP server process
# - Spawns the server process with specified parameters
# - Establishes JSON-RPC communication channel
# - Handles server lifecycle (start/stop) automatically
# - 60-second timeout prevents hanging connections
async with MCPServerStdio(params=fetch_params, client_session_timeout_seconds=60) as server:
    # Query the server for available tools
    # This is the "discovery" phase where we learn what capabilities the server provides
    # The fetch server typically provides tools like:
    # - fetch(): Get content from web pages
    # - search(): Search the web
    # - screenshot(): Take screenshots of web pages
    fetch_tools = await server.list_tools()

# Display the available tools from the fetch MCP server
# This shows us what web browsing capabilities our AI agent can access
fetch_tools
```

## And now 2 more MCP servers - this time Javascript based, with node

### If the next cell hangs, see the important troubleshooting at the end of the SETUP-node instructions linked above.

```python
# MCP CLIENT SETUP: Connect to Playwright MCP Server
# This demonstrates connecting to a JavaScript-based MCP server using Node.js

# Configure parameters for Playwright MCP server
# "npx" is Node.js package runner that executes JavaScript MCP servers
# "@playwright/mcp@latest" provides advanced web automation tools
playwright_params = {"command": "npx","args": [ "@playwright/mcp@latest"]}

# Create MCP Client connection to Playwright server
# This server provides powerful browser automation capabilities beyond basic fetching
async with MCPServerStdio(params=playwright_params, client_session_timeout_seconds=60) as server:
    # Discover available tools from Playwright server
    # Expected tools include: navigate, click, fill forms, take screenshots, wait for elements
    playwright_tools = await server.list_tools()

# Display Playwright tools - these enable sophisticated web interactions
# Unlike fetch server, Playwright can interact with dynamic web pages and SPAs
playwright_tools
```

```python
# MCP CLIENT SETUP: Connect to Filesystem MCP Server
# This server provides secure file operations within a sandboxed directory

# Create absolute path to sandbox directory for secure file operations
# This restricts file access to only the sandbox folder for security
sandbox_path = os.path.abspath(os.path.join(os.getcwd(), "sandbox"))

# Configure parameters for filesystem MCP server
# "-y" flag auto-accepts npm prompts, "@modelcontextprotocol/server-filesystem" is the official filesystem server
files_params = {"command": "npx", "args": ["-y", "@modelcontextprotocol/server-filesystem", sandbox_path]}

# Create MCP Client connection to filesystem server
# This server provides safe file operations limited to the sandbox directory
async with MCPServerStdio(params=files_params,client_session_timeout_seconds=60) as server:
    # Discover available file operation tools
    # Expected tools: read_file, write_file, create_directory, list_directory, etc.
    file_tools = await server.list_tools()

# Display filesystem tools - these enable AI agents to work with files safely
# All operations are restricted to the sandbox path for security
file_tools
```

### And now.. bring on the Agent with Tools!

```python
# AGENT SETUP: Define instructions for web browsing behavior
# These instructions guide the AI agent on how to navigate websites effectively
instructions = """
You browse the internet to accomplish your instructions.
You are highly capable at browsing the internet independently to accomplish your task, 
including accepting all cookies and clicking 'not now' as
appropriate to get to the content you need. If one website isn't fruitful, try another. 
Be persistent until you have solved your assignment,
trying different options and sites as needed.
"""

# MULTI-MCP AGENT: Create agent with both filesystem and browser capabilities
# This demonstrates MCP's power - combining multiple servers for complex tasks
async with MCPServerStdio(params=files_params, client_session_timeout_seconds=60) as mcp_server_files:
    async with MCPServerStdio(params=playwright_params, client_session_timeout_seconds=60) as mcp_server_browser:
        # Create AI agent with multiple MCP servers
        # Agent can now browse web AND write files - powerful combination!
        agent = Agent(
            name="investigator", 
            instructions=instructions, 
            model="gpt-4.1-mini",
            mcp_servers=[mcp_server_files, mcp_server_browser]  # Multiple MCP servers
            )
        # Execute task with tracing enabled for debugging
        # Task combines web browsing (Playwright) + file writing (filesystem)
        with trace("investigate"):
            result = await Runner.run(agent, "Find a great recipe for Banoffee Pie, then summarize it in markdown to banoffee.md")
            print(result.final_output)
```
