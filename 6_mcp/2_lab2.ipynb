### Now we write an MCP server and use it directly!

```python
# Now let's use our accounts server as an MCP server


params = {"command": "uv", "args": ["run", "accounts_server.py"]}
async with MCPServerStdio(params=params, client_session_timeout_seconds=30) as server:
    mcp_tools = await server.list_tools()
    display(Markdown(f"### MCP Tools:\n\n{mcp_tools}"))
```

```python
mcp_tools
```

```python
instructions = "You are able to manage an account for a client, and answer questions about the account."
request = "My name is Ed and my account is under the name Ed. What's my balance and my holdings?"
model = "gpt-4.1-mini"
```

```python

async with MCPServerStdio(params=params, client_session_timeout_seconds=30) as mcp_server:
    agent = Agent(name="account_manager", instructions=instructions, model=model, mcp_servers=[mcp_server])
    with trace("account_manager"):
        result = await Runner.run(agent, request)
    display(Markdown(result.final_output))
```

### Now let's build our own MCP Client

### Now let's build our own MCP Client

The key insight here is that we're creating a **bridge** between two different tool systems:

1. **MCP Tools** - What your server provides (get_balance, buy_shares, etc.)
2. **OpenAI Tools** - What the OpenAI Agents SDK expects

The `get_accounts_tools_openai()` function does this conversion automatically. When an OpenAI agent calls a tool, it actually calls back to your MCP server under the hood.

This is **not** how OpenAI tools work internally - it's a custom adapter for interoperability!

```python
# Now let's build our own MCP Client

# Step 1: Get raw MCP tools from the server
from accounts_client import get_accounts_tools_openai, read_accounts_resource, list_accounts_tools

mcp_tools = await list_accounts_tools()
print("Raw MCP Tools:")
print(mcp_tools)

# Step 2: Convert MCP tools to OpenAI-compatible format
openai_tools = await get_accounts_tools_openai()
print("\nConverted to OpenAI Tools:")
print(openai_tools)
```

### Understanding the MCP ↔ OpenAI Bridge

The confusion often comes from thinking this is how OpenAI tools work internally. **It's not!**

This code demonstrates **interoperability** - connecting two different tool ecosystems:

**MCP Server** (your accounts_server.py) → **MCP Client** (accounts_client.py) → **OpenAI Agent**

When you call `get_accounts_tools_openai()`, you're not getting "real" OpenAI tools. You're getting **wrapper objects** that:
1. Look like OpenAI tools to the Agents SDK
2. Actually call your MCP server behind the scenes

This is powerful for integration but can be confusing because it's not OpenAI's native tool system!

```python
request = "My name is Ed and my account is under the name Ed. What's my balance?"

with trace("account_mcp_client"):
    agent = Agent(name="account_manager", instructions=instructions, model=model, tools=openai_tools)
    result = await Runner.run(agent, request)
    display(Markdown(result.final_output))
```
